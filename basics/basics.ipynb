**Question 17:** Write a Python program to check if a year is a leap year.

Leap Year Rules
A year is a leap year if:

Divisible by 4 → It’s a leap year unless:

Divisible by 100 → Then it’s not a leap year unless:

Divisible by 400 → Then it is a leap year.

```python
# Your code here
year=int(input("Enter a year: "))
if year%400==0 or (year%4==0 and year%100 !=0):
    print(year," is a leap year")
else:
    print(year, " is not a leap year")
```

**Question 18:** Write a Python program to reverse a string.

```python
# Your code here
country=input("Enter a country name:")
temp=""
for i in range(len(country)-1,-1,-1):
    temp+=country[i]
print(temp)

temp=""
for i in range(0,len(country)):
    temp=country[i]+temp
print(temp)
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
# Your code here
s=input("Enter a string:")
check_palindrome=""
for i in range(0,len(s)):
    check_palindrome=s[i]+check_palindrome
if s==check_palindrome:
    print(s," is a palindrome")
else:
    print(s," is not a palindrome")


s1=s[::-1]
if s==s1:
    print("palindrome")
else:
    print("Not a palindrome")
```

**Question 20:** Write a Python program to sort a list of numbers in ascending order.

```python
# Your code here
numbers=[5,3,7,2,9,11,8,1,4,0]
# numbers=[15,9,5,20,1,8,4,3]
for i in range(0,len(numbers)):
    for j in range(i+1,len(numbers)):
        if numbers[i] > numbers[j]:
            numbers[i],numbers[j]=numbers[j],numbers[i]
print(numbers)
```
