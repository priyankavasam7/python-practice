Write a program that asks the user to input a number and prints whether the number is positive or negative.

```python
n=int(input("Enter a number: "))
if n>0:
    print(f"{n} is positive")
elif n==0:
    print(f"{n} is zero")
else:
    print(f"{n} is negative")
```

Write a program that asks the user to input a number and prints whether the number is positive and even, positive and odd, or negative.

```python
n=int(input("Enter a non zero number: "))
if n>0:
    if n%2==0:
        print(f"{n} is positive even number")
    else:
        print(f"{n} is positive odd number")
else:
    print(f"{n} is negative number")
```

Write a program that asks the user to input a number and prints all the even numbers from 1 to that number using a for loop.

```python
n=int(input("Enter a number (to get all even numbers upto that number): "))
s=""
for i in range(1,n+1):
    if i%2==0:
        s+=str(i)+" "
print(f"Even numbers from 1 to {n} are: {s}")
```

Write a program that calculates the factorial of a number input by the user using a while loop.

```python
n=int(input("Enter a number: (to find factorial)"))
if n==0:
    print(1)
else:
    fact=1
    while n!=0:
        fact*=n
        n-=1
    print(f"Factorial of the number is {fact}")
```

Write a program that checks if a number input by the user is a prime number using a for loop.

```python
n=int(input("Enter number to check if it is prime "))
if n<=0:
    print("Prime numbers are natural numbers, enter a number greater than zero")
elif n==1:
    print("1 is neither prime nor composite")
else:
    for i in range(2,n):
        if n%i==0:
            print(f"{n} is not a prime number")
            break 
    else:
        print(f"{n} is a prime number")
```

Write a program that prints the first n Fibonacci numbers, where n is input by the user.

```python
n=int(input("enter a number:"))
first_number=0
second_number=1
print(first_number)
print(second_number)
while n>2:
    next_number=first_number+second_number
    print(next_number)
    first_number=second_number
    second_number=next_number
    n-=1
```
